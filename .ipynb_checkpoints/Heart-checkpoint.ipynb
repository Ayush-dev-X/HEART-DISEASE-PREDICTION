## Data Preprocessing and Cleaning

```python
df_encoded = pd.get_dummies(df,drop_first=True)
```

```python
df_encoded
```

```python
df_encoded.columns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = df_encoded.drop('HeartDisease',axis=1)
y = df_encoded['HeartDisease']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test) 
```

```python
knn = KNeighborsClassifier()

knn.fit(X_train_scaled, y_train)
```

```python
manual_data = pd.DataFrame([{
    'Age': 35,
    'RestingBP': 120,
    'Cholesterol': 180,
    'FastingBS': 0,
    'MaxHR': 175,
    'Oldpeak': 0.0,
    'FamilyHistory': 0,
    'Lpa_Level': 0,
    'FH_Mutation': 0,
    'Genetic_Risk_Score': 0,

    'HeartDisease': 0,

    'Sex_M': 0,

    'ChestPainType_ATA': 1,
    'ChestPainType_NAP': 0,
    'ChestPainType_TA': 0,

    'RestingECG_Normal': 1,
    'RestingECG_ST': 0,

    'ExerciseAngina_Y': 0,

    'ST_Slope_Flat': 0,
    'ST_Slope_Up': 1,

    'rs10757274_Genotype_AG': 0,
    'rs10757274_Genotype_GG': 0,

    'APOE_Genotype_e3/e3': 1,
    'APOE_Genotype_e3/e4': 0,
    'APOE_Genotype_e4/e4': 0
}])

# Make sure columns are in exactly the same order as training data
manual_data = manual_data[X.columns]

# Scale the input
manual_scaled = scaler.transform(manual_data)

# Prediction
prediction = knn.predict(manual_scaled)

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Heart Disease: YES")
else:
    print("Heart Disease: NO")
```

```python
models = {
    "Logistic Regression":LogisticRegression(), 
    "KNN":KNeighborsClassifier(), 
    "Native Byeas":GaussianNB(), 
    "Decision tree":DecisionTreeClassifier(), 
    "SVM":SVC(probability=True)
}
```

```python
results = []
```

```python
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    results.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'F1 Score': round(f1, 4)
    })
```
