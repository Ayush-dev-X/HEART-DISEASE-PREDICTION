```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = df_encoded.drop('HeartDisease',axis=1)
y = df_encoded['HeartDisease']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test) 
```

```python
knn = KNeighborsClassifier()

knn.fit(X_train_scaled, y_train)
```

```python
manual_data = pd.DataFrame([{
    'Age': 35,
    'RestingBP': 120,
    'Cholesterol': 180,
    'FastingBS': 0,
    'MaxHR': 175,
    'Oldpeak': 0.0,
    'Sex_M': 0,
    'ChestPainType_ATA': 1,
    'ChestPainType_NAP': 0,
    'ChestPainType_TA': 0,
    'RestingECG_Normal': 1,
    'RestingECG_ST': 0,
    'ExerciseAngina_Y': 0,
    'ST_Slope_Flat': 0,
    'ST_Slope_Up': 1
}])

manual_data = manual_data[X.columns]

manual_scaled = scaler.transform(manual_data)

prediction = knn.predict(manual_scaled)

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Heart Disease: YES")
else:
    print("Heart Disease: NO")
```

```python
models = {
    "Logistic Regression":LogisticRegression(), 
    "KNN":KNeighborsClassifier(), 
    "Native Byeas":GaussianNB(), 
    "Decision tree":DecisionTreeClassifier(), 
    "SVM":SVC(probability=True)
}
```

```python
results = []
```

```python
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    results.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'F1 Score': round(f1, 4)
    })
```

```python
# ============================================================
# CARDIOSENSE - COMPLETE FLASK BACKEND
# ============================================================

from flask import Flask, render_template, request
import pandas as pd
import numpy as np
import joblib
import os


# ============================================================
# 1. CREATE FLASK APP
# ============================================================

app = Flask(__name__)


# ============================================================
# 2. LOAD TRAINED MODEL + PREPROCESSING
# ============================================================

MODEL_FILE = "logistic_regression_heart.pkl"
SCALER_FILE = "scaler.pkl"
COLUMNS_FILE = "columns.pkl"


# Check files exist
for file in [MODEL_FILE, SCALER_FILE, COLUMNS_FILE]:

    if not os.path.exists(file):
        raise FileNotFoundError(
            f"Required file not found: {file}"
        )


# Load trained Logistic Regression
model = joblib.load(MODEL_FILE)

# Load scaler
scaler = joblib.load(SCALER_FILE)

# Load exact training columns
columns = joblib.load(COLUMNS_FILE)


# ============================================================
# 3. VERIFY MODEL
# ============================================================

print()
print("==============================================")
print("        CARDIO.SENSE MODEL STATUS")
print("==============================================")

print("Model        :", type(model).__name__)
print("Fitted       :", hasattr(model, "coef_"))
print("Features     :", len(columns))
print("Scaler       :", type(scaler).__name__)

if hasattr(model, "n_features_in_"):
    print("Model inputs :", model.n_features_in_)

if hasattr(scaler, "n_features_in_"):
    print("Scaler inputs:", scaler.n_features_in_)

print("==============================================")
print()


# ============================================================
# 4. HOME PAGE
# ============================================================

@app.route("/")
def home():

    return render_template("index.html")


# ============================================================
# 5. PREDICTION PAGE
# ============================================================

@app.route("/predict", methods=["GET", "POST"])
def predict():

    # --------------------------------------------------------
    # GET REQUEST
    # --------------------------------------------------------

    if request.method == "GET":

        return render_template("predict.html")


    # --------------------------------------------------------
    # POST REQUEST
    # --------------------------------------------------------

    try:

        # ====================================================
        # READ FORM VALUES
        # ====================================================

        age = float(request.form["age"])

        sex = request.form["sex"]

        chest_pain = request.form["cp"]

        resting_bp = float(
            request.form["trestbps"]
        )

        cholesterol = float(
            request.form["chol"]
        )

        fasting_bs = int(
            request.form["fbs"]
        )

        resting_ecg = request.form["restecg"]

        max_hr = float(
            request.form["thalach"]
        )

        exercise_angina = request.form["exang"]

        oldpeak = float(
            request.form["oldpeak"]
        )

        st_slope = request.form["slope"]


        # ====================================================
        # CREATE EXACT 15 FEATURES USED DURING TRAINING
        # ====================================================

        input_data = pd.DataFrame([{

            # Numerical features
            "Age": age,

            "RestingBP": resting_bp,

            "Cholesterol": cholesterol,

            "FastingBS": fasting_bs,

            "MaxHR": max_hr,

            "Oldpeak": oldpeak,


            # Sex
            "Sex_M":
                1 if sex == "M" else 0,


            # Chest Pain
            "ChestPainType_ATA":
                1 if chest_pain == "ATA" else 0,

            "ChestPainType_NAP":
                1 if chest_pain == "NAP" else 0,

            "ChestPainType_TA":
                1 if chest_pain == "TA" else 0,


            # Resting ECG
            "RestingECG_Normal":
                1 if resting_ecg == "Normal" else 0,

            "RestingECG_ST":
                1 if resting_ecg == "ST" else 0,


            # Exercise Angina
            "ExerciseAngina_Y":
                1 if exercise_angina == "Y" else 0,


            # ST Slope
            "ST_Slope_Flat":
                1 if st_slope == "Flat" else 0,

            "ST_Slope_Up":
                1 if st_slope == "Up" else 0

        }])


        # ====================================================
        # FORCE EXACT COLUMN ORDER
        # ====================================================

        input_data = input_data.reindex(
            columns=columns,
            fill_value=0
        )


        # ====================================================
        # SAFETY CHECK
        # ====================================================

        if input_data.shape[1] != len(columns):

            raise ValueError(
                f"Feature mismatch: model expects "
                f"{len(columns)} features but received "
                f"{input_data.shape[1]}"
            )


        # ====================================================
        # SCALE INPUT
        # ====================================================

        input_scaled = scaler.transform(
            input_data
        )


        # ====================================================
        # MODEL PREDICTION
        # ====================================================

        prediction = int(
            model.predict(input_scaled)[0]
        )


        # ====================================================
        # MODEL PROBABILITY
        # ====================================================

        if hasattr(model, "predict_proba"):

            probabilities = model.predict_proba(
                input_scaled
            )[0]

            # Probability of class 1
            risk_probability = float(
                probabilities[1]
            )

            risk_percentage = round(
                risk_probability * 100,
                1
            )

        else:

            risk_percentage = (
                100.0 if prediction == 1
                else 0.0
            )


        # ====================================================
        # RISK CATEGORY
        # ====================================================

        if risk_percentage < 30:

            risk_level = "Low Risk"

        elif risk_percentage < 60:

            risk_level = "Moderate Risk"

        else:

            risk_level = "High Risk"


        # ====================================================
        # PREDICTION LABEL
        # ====================================================

        if prediction == 1:

            prediction_label = (
                "Higher likelihood detected"
            )

        else:

            prediction_label = (
                "Lower likelihood detected"
            )


        # ====================================================
        # PRINT RESULT IN JUPYTER CONSOLE
        # ====================================================

        print()
        print("----------------------------------------------")
        print("CARDIOSENSE PREDICTION")
        print("----------------------------------------------")
        print("Age             :", age)
        print("Sex             :", sex)
        print("Chest Pain      :", chest_pain)
        print("Resting BP      :", resting_bp)
        print("Cholesterol     :", cholesterol)
        print("Fasting BS      :", fasting_bs)
        print("Resting ECG     :", resting_ecg)
        print("Maximum HR      :", max_hr)
        print("Exercise Angina :", exercise_angina)
        print("Oldpeak         :", oldpeak)
        print("ST Slope        :", st_slope)
        print("----------------------------------------------")
        print("Prediction      :", prediction)
        print("Risk Percentage :", risk_percentage, "%")
        print("Risk Level      :", risk_level)
        print("----------------------------------------------")


        # ====================================================
        # SEND RESULT TO RESULT.HTML
        # ====================================================

        return render_template(

            "result.html",

            prediction=prediction,

            prediction_label=prediction_label,

            risk_percentage=risk_percentage,

            risk_level=risk_level,

            age=age,

            sex=sex,

            chest_pain=chest_pain,

            resting_bp=resting_bp,

            cholesterol=cholesterol,

            fasting_bs=fasting_bs,

            resting_ecg=resting_ecg,

            max_hr=max_hr,

            exercise_angina=exercise_angina,

            oldpeak=oldpeak,

            st_slope=st_slope

        )


    # ========================================================
    # ERROR HANDLING
    # ========================================================

    except Exception as e:

        print()
        print("==============================================")
        print("CARDIOSENSE PREDICTION ERROR")
        print("==============================================")
        print(type(e).__name__, ":", e)
        print("==============================================")

        return render_template(
            "predict.html",
            error=str(e)
        )


# ============================================================
# 6. RUN CARDIOSENSE
# ============================================================

if __name__ == "__main__":

    print()
    print("==============================================")
    print("       CARDIO.SENSE IS ONLINE")
    print("==============================================")
    print("Open: http://127.0.0.1:5000")
    print("==============================================")
    print()

    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False,
        use_reloader=False
    )
```
